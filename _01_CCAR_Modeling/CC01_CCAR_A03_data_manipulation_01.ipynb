# Step 2: Upload two types of CSV files

```python
my_2_col = ['mon', 'yr']
```

### Step 2-A: Upload simulated CCAR file

```python
df_simu= pd.read_csv( myout + "/CCAR_simulated_data_20260318_01.csv" )

df_simu= df_simu.drop( columns= ['Unnamed: 0'])

df_simu[['yr', 'mon', 'day']] = df_simu['report_date'].str.split('-', expand=True)
df_simu['yrmo'] = df_simu['yr'] + df_simu['mon']
df_simu['report_yrmo'] = df_simu['yrmo']

for i in my_2_col:
    print( "---------2-A-1 Freq Table of {}".format(i), df_simu[i].value_counts() )

df_simu= df_simu.drop( columns = ['mon', 'yr' , 'day'])

print( "\n---------2-A-3 Freq Table of yrmo", df_simu.yrmo.value_counts() )

print( "\n---------2-A-3 Freq Table of product_type", df_simu.product_type.value_counts() )

df_simu[['_yr', '_mon', '_day']] = df_simu['default_date'].str.split('-', expand=True)
df_simu['default_yrmo'] = df_simu['_yr'] + df_simu['_mon']

df_simu['flag_default'] = 0
df_simu.loc[ df_simu['default_yrmo'] <= df_simu['report_yrmo'], 'flag_default' ] = 1

df_simu = df_simu.drop(columns= ['_yr', '_mon', '_day' ] )

print( "\n---------2-A-4 Freq Table of flag_default", df_simu.flag_default.value_counts() )

print( "\n---------2-A-5 Freq Table of default_yrmo", df_simu.product_type.value_counts() )

df_simu.info()
```

### 2-B: Upload CCAR Base scenario file

```python
month_dic= {'Jan' : '01', 'Feb' : '02', 'Mar' : '03',  'Apr' : '04', 'May' : '05', 'Jun' : '06',
         'Jul' : '07', 'Aug' : '08', 'Sep' : '09',  'Oct' : '10', 'Nov' : '11', 'Dec' : '12' }

for item, key in month_dic.items():
  print( item, key )
```

```python
df_base= pd.read_csv( myout + "/basereg.csv" )

# df_base['yrmo'] = pd.to_datetime( df_base['nwdate'],  format='%b-%d')
# This solution does not work since it will convert Apr-01 to 1900-04-01, which is completely wrong.

df_base[[ 'mon', 'yr']]= df_base['nwdate'].str.split( '-', expand=True )

for i in my_2_col:
    print( "---------2-B-1 Freq Table of {} ---\n".format(i), df_base[i].value_counts() )

# Create the new column
df_base['month'] = df_base['mon'].map(month_dic)

print( "\n---------2-B-3 Freq Table of month ---\n", df_base.month.value_counts() )

df_base['yrmo'] = '20' + df_base[ 'yr'] + df_base['month']

print( "\n---------2-B-4 Freq Table of yrmo ---\n", df_base.yrmo.value_counts() )

# print( "---------2-B-5 Freq Table of nwdate ---\n", df_base.nwdate.value_counts() )

print( "\n---------2-B-6 Freq Table of nwdate /yrmo ---\n", df_base.groupby( ['nwdate', 'yrmo'] ).size() )

df_base= df_base.drop( columns = ['mon', 'yr' , 'month'])

df_base.info()
```

```python
df_base_1= df_base[['yrmo', 'delta_Unemployment1', 'delta_Mortgage_rate1', 'delta_House_Price_Index__Level1',
                            'delta_Unemployment3', 'delta_Mortgage_rate3', 'delta_House_Price_Index__Level3',
                            'delta_Unemployment6', 'delta_Mortgage_rate6', 'delta_House_Price_Index__Level6',
                            'delta_Unemployment12', 'delta_Mortgage_rate12', 'delta_House_Price_Index__Level12',
                            'delta_Unemployment24', 'delta_Mortgage_rate24', 'delta_House_Price_Index__Level24'   ]]

df_base_1.info()
```

# Step 3: Combine two types of input files together

```python
df_simu_1= pd.merge(df_simu, df_base_1, on='yrmo', how='inner', indicator='flag_merge')

print ( df_simu_1.flag_merge.value_counts() )

print(pd.crosstab(index= df_simu_1['yrmo'], columns=df_simu_1['product_type'] ) )
```
